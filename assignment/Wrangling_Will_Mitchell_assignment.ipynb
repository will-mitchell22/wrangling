1. This paper is about the process of tidying data and making it more clean and efficient to manipulate and visualize.  The paper describes how vast quantities of effort are put into cleaning data, but there has been little work done to make this process more efficient.  The paper describes tools and effective ways to make data sets more tidy and easier to work with.

2. The "tidy data standard" is intended to create a much easier way to clean data, without having to "reinvent the wheel every time".  The "tidy data standard" allows for understanding and exploration of the data, and creating a simple and cohesive process that uses tools to efficiently clean the data.  The "tidy data standard" can be applied to varying data sets of different "messiness" levels, and can have a similar process using similar tools to make the cleaning job much easier.

3. The sentence "Like families, tidy datasets are all alike but every messy dataset is messy in its own way", means that when datasets are tidy, they will all follow a similar structure/plan to one another, and skills used to analyze one will be easily transferrable to the next.  Messy datasets on the other hand, are all different in their own way because there is no set way that a data set can be messy, leading to problem solving from Data Scientists to clean the data.  The sentence "For a given dataset, it's usually easy to figure out what are observations and what are variables, but it is surprisingly difficult to precisely define values, variables, and observations", means that when you have variables and observations, it is easy to tell which are which, but in certain instances, defining the variables and observations themselves can be confusing.  The reading gives an example of hight and width both being values of a dimension variable, showing how these variables can now be linked together under one.

4. Wickham defines values as as either numbers or strings, organized in two ways where they beling to a variable and an observation.  Variables contain all values that measure the same underlying attribute across units.  Observations contain all the values measured on the same unit across attributes.

5. "Tidy Data" is defined in section 2.3 as a standard way of mapping the meaning of a dataset to its structure.  Within a tidy dataset, there each variable forms a column, each observation forms a row, and each type of observational unit forms a table.  Messy data is also described as any sort of deviation from this alignment.

6. The five most common problems with messy datasets are, that column headers are values, not variable names, multiple variables are stored in one column, variables are stored in both rows and columns, multiple types of observational units are stored in the same table, and that a single observational unit is stored in multiple tables.  The data in table 4 is messy data because this table is designed for presentation, but the column headers are values, not variable names and variables form both the rows and the columns.  Melting a dataset means to essentially stack it by turning columns into rows.

7. Table 11 is messy because it has variables stored in both rows and columns.  It has variables in individual colums, spread across columns, and across rows.  Table 12 is tidy and molten by having an almost tidy dataset, but the element column contains the names of variables, and then the tidy data having each row representing measurements for a single day.

8. The "chicken-and-egg" problem with focusing on tidy data is that if tidy data is only as useful as the tools that work with it, then tidy tools will be inextricably linked to tidy data.  Wickham hopes that in the future others will build on the tidy data framework to develop better data storage strategies and better tools. Wickham hopes that in the future to use methodologies from varying fields to improve understanding in the cognitive side of data analysis and improve design abilities.  Wickham also stated that we could consider tidy tools that can automatically choose between array-tidy and dataframe-tidy formats to optimize memory usage and performance.

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

```python
df = pd.read_csv('/content/wrangling/assignment/data/airbnb_hw.csv')
print(df.shape, '\n')
df.head()
```

```python
price = df['Price']
price.unique()
```

```python
price = df['Price']
price = price.str.replace(',','.') # This will replace the commas with periods.
print(price.unique() , '\n')
price = pd.to_numeric(price,errors='coerce') # This will change price to a float/numeric type of variable
print(price.unique(), '\n')
print('Total missing: ', sum(price.isnull())) # This will count the number of missing values
```

```python
df = pd.read_csv('/content/wrangling/assignment/data/sharks.csv', low_memory=False)
df.head()
df.columns.tolist() # These two functions will show the dataframe and convert them to columns
```

```python
df['Type'].value_counts() # This counts values
```

```python
type = df['Type'] #This allows for the Type variable to have values
type = type.replace(['Sea Disaster', 'Boat', 'Boating', 'Boatomg'], 'Watercraft') #This is all of the values related to watersports removed being added to Watercraft
type.value_counts()
type = type.replace(['Questionable', 'Unconfirmed', 'Unverified', 'Under investigation'],np.nan) # This disregards these values
type.value_counts()
df['Type'] = type # Allows for a clean version
del type # Then destroys the temporary 'Type'
df['Type'].value_counts()
```

```python
df = pd.read_parquet('/content/wrangling/data/justice_data.parquet')
df.head()
df.columns.tolist() # These will display the dataframe in a cleaner format
```

```python
release = df['WhetherDefendantWasReleasedPretrial']
print(release.unique(), '\n')
print(release.value_counts(), '\n')
release = release.replace(9,np.nan) #9's were a number that should be replaced to np.nan
print(release.value_counts(),'\n')
sum(release.isnull()) # This shows the number of values that are still missing
df['WhetherDefendantWasReleasedPretrial'] = release # This shows your results with missing values cleared
del release
```

```python
length = df['ImposedSentenceAllChargeInContactEvent']
type = df['SentenceTypeAllChargesAtConvictionInContactEvent']
print(length.unique(), '\n') # This shows values that are missing
print(type.unique()) # This shows the values in the Type
length = pd.to_numeric(length,errors='coerce') # This changes to a numeric variable
length_NA = length.isnull() # This will create a dummy variable to represent missing values
print(np.sum(length_NA), '\n') #This shows the number of missing values at 274
print(pd.crosstab(length_NA,type), '\n') #Although the number of missing values was already reduced to 274, these next steps will show how to have reduced to this from a higher number of missing values
# This will examine cases that should be dismissed
length = length.mask(type == 4, 0) # This looks at the cases where charges were dismissed and replaces the length with 0
length = length.mask(type == 9, np.nan) # This will make the length np.nan when the charges are unclear
length_NA = length.isnull() # This creates another dummy variable to represent missing values
print(pd.crosstab(length_NA, type), '\n')
print(np.sum(length_NA),'\n') # This will show a new number of missing values (which should be 274)
df['ImposedSentenceAllChargeInContactEvent'] = length
del length, type
```
